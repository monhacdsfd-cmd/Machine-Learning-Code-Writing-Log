# 1.Linear Regression
### 공부 시간에 따른 시험 점수

```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv('LinearRegressionData.csv')
```

```python
dataset.head() #head : 상위 5개의 데이터만 출력
```

```python
X = dataset.iloc[:, :-1].values #처음부터 마지막 캄럼 직전까지의 데이터 : 독립 변수-원인(hour)
y = dataset.iloc[:, -1].values #마지막 칼럼 데이터 종속 변수-결과 (score)
```

```python
X,y
```

```python
from sklearn.linear_model import LinearRegression
reg = LinearRegression() #객체 생성
reg.fit(X,y) #학습 (모델 생성)
```

```python
y_pred = reg.predict(X) #X 에 대한 에측 값
y_pred
```

```python
plt.scatter(X,y, color='blue') #산점도
plt.plot(X, y_pred, color='green') #선 그래프
plt.title('Hours vs Scores') #제목
plt.xlabel('Hours') #x축 이름
plt.ylabel('Scores') #y축 이름
plt.show()
```

```python
if_hours = float(input("공부한 시간을 입력하세요 : "))
```

```python
print(f"{if_hours}시간 공부했을 때 예상 점수: {reg.predict([[if_hours]])}") #if_hours시간 공부했을 때 예상 점수
```
